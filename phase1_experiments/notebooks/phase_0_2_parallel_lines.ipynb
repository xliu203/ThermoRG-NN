# Phase 0.2: BN vs LN Parallel Lines Core Test
## Measure β(γ) for BN and LN, check slope equality

**Protocol:** ConvNet L=5, D ∈ {64, 256, 1024}, BN vs LN, 6 learning rates, 3 seeds, 200 epochs

This notebook tests the core hypothesis that BN and LN should have the same slope β(ln(γ)) - the parallel lines hypothesis from ThermoRG theory.

**What to expect:**
- BN and LN data points should fall on parallel lines (same slope m, different intercept c)
- ANCOVA test checks whether slopes are statistically equal
